```python
# --- Jupyter Notebook Script: Dynamic Offset Track Generator ---
#
# INSTRUCTIONS:
# 1. Ensure 'gateway-4-telemetry.csv' is in the same directory as this notebook.
# 2. Run this cell.
# 3. Adjust the 'MAX_TRACK_WIDTH_M' and 'MIN_OFFSET_RATIO' parameters below to fine-tune the result.
#

# Load necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def smooth_circular(df, window_size):
    """Applies a circular rolling mean to latitude and longitude columns."""
    if len(df) <= window_size:
        return df # Cannot smooth if data size is too small

    # Create a cyclic series by concatenating the start/end segments
    pre_pad = df.iloc[-window_size//2:]
    post_pad = df.iloc[:window_size//2]
    
    df_padded = pd.concat([pre_pad, df, post_pad], ignore_index=True)
    
    # Apply rolling mean to the padded series
    df_padded['latitude'] = df_padded['latitude'].rolling(window=window_size, center=True).mean()
    df_padded['longitude'] = df_padded['longitude'].rolling(window=window_size, center=True).mean()
    
    # Slice back to the original length
    start_index = window_size // 2
    end_index = start_index + len(df)
    
    return df_padded.iloc[start_index:end_index].reset_index(drop=True)


def create_non_overlapping_boundaries(df_centerline, max_track_width=10, min_offset_ratio=0.7, smoothing_window=50):
    """
    Creates parallel track boundaries that avoid overlapping in tight corners
    by dynamically adjusting the offset based on local track curvature.
    
    NOTE: This version uses the configuration that yielded a good result:
    INNER_RATIO = 0.8 (Inner Boundary gets max 8m offset, used for the smoother line)
    OUTSIDE_RATIO = 0.2 (Outer Boundary gets max 2m offset, used for the noisier line)
    The plotting labels have been corrected to match the requested 2m/8m split.

    Args:
        df_centerline (pd.DataFrame): DataFrame with 'latitude' and 'longitude' of the centerline.
        max_track_width (int): The maximum desired width of the track in meters. (Must be 10m for split).
        min_offset_ratio (float): The minimum ratio of the offset at the tightest corner
                                  (e.g., 0.7 means the width can be reduced to 70% of max).
        smoothing_window (int): Size of the rolling mean window to apply to the final boundary coordinates.

    Returns:
        tuple: A tuple containing two DataFrames: 
               (outer_boundary_df, inner_boundary_df) with latitude/longitude.
    """
    if df_centerline.empty:
        raise ValueError("Centerline data is empty. Cannot generate boundaries.")

    # Define asymmetric offset ratios based on the user's successful switch:
    # We assume 'Inner Boundary' in the plot (red line in your image) corresponds to 8m, 
    # and 'Outer Boundary' (blue line in your image) corresponds to 2m.
    # The actual geometric side (positive/negative normal) determines which line is which.
    INNER_8M_RATIO = 0.2 
    OUTER_2M_RATIO = 0.8 
    
    # 1. Coordinate Conversion (Lat/Lon to Local Meters)
    start_lat = df_centerline.iloc[0]['latitude']
    start_lon = df_centerline.iloc[0]['longitude']
    
    LAT_M_PER_DEGREE = 111320.0
    
    # Conversion factor for longitude changes based on latitude (it's a Series)
    LON_M_PER_DEGREE_SERIES = LAT_M_PER_DEGREE * np.cos(np.radians(df_centerline['latitude']))

    # Store the single conversion factor at the START point for consistent reverse projection
    LON_M_PER_DEGREE_START = LON_M_PER_DEGREE_SERIES.iloc[0] 

    df_centerline['x_m'] = (df_centerline['longitude'] - start_lon) * LON_M_PER_DEGREE_SERIES
    df_centerline['y_m'] = (df_centerline['latitude'] - start_lat) * LAT_M_PER_DEGREE

    # 2. Curvature Calculation
    # Calculate first and second derivatives for curvature formula
    dx = np.gradient(df_centerline['x_m'])
    dy = np.gradient(df_centerline['y_m'])
    d2x = np.gradient(dx)
    d2y = np.gradient(dy)
    
    # Curvature formula: k = |x'y'' - y'x''| / (x'^2 + y'^2)^(3/2)
    curvature = np.abs(dx * d2y - dy * d2x) / ((dx**2 + dy**2)**1.5)
    
    # 3. Dynamic Offset Scaling
    if curvature.max() == curvature.min():
        normalized_curvature = np.zeros_like(curvature)
    else:
        normalized_curvature = (curvature - curvature.min()) / (curvature.max() - curvature.min())
    
    # Calculate the scale factor: 1.0 (straight) down to min_offset_ratio (tightest corner)
    offset_scale = (1 - normalized_curvature) * (1.0 - min_offset_ratio) + min_offset_ratio
    
    # Calculate asymmetric dynamic offsets
    dynamic_offset_8m = (max_track_width * INNER_8M_RATIO) * offset_scale
    dynamic_offset_2m = (max_track_width * OUTER_2M_RATIO) * offset_scale

    # 4. Normal Vector Calculation
    magnitude = np.sqrt(dx**2 + dy**2)
    dx_norm = dx / magnitude
    dy_norm = dy / magnitude
    
    # Normal vector is perpendicular to the tangent ([-dy_norm, dx_norm])
    nx, ny = -dy_norm, dx_norm

    # 5. Calculate Offset Points
    # Boundary 1 (Corresponds to 8m offset, mapped to red line in your image)
    b1_x = df_centerline['x_m'] + nx * dynamic_offset_8m
    b1_y = df_centerline['y_m'] + ny * dynamic_offset_8m
    
    # Boundary 2 (Corresponds to 2m offset, mapped to blue line in your image)
    b2_x = df_centerline['x_m'] - nx * dynamic_offset_2m
    b2_y = df_centerline['y_m'] - ny * dynamic_offset_2m
    
    # 6. Convert back to Lat/Lon
    b1_df = pd.DataFrame({
        'latitude': start_lat + b1_y / LAT_M_PER_DEGREE,
        'longitude': start_lon + b1_x / LON_M_PER_DEGREE_START
    })
    
    b2_df = pd.DataFrame({
        'latitude': start_lat + b2_y / LAT_M_PER_DEGREE,
        'longitude': start_lon + b2_x / LON_M_PER_DEGREE_START
    })

    # 7. Post-processing Smoothing (using circular rolling mean)
    if smoothing_window > 1:
        b1_df = smooth_circular(b1_df, smoothing_window)
        b2_df = smooth_circular(b2_df, smoothing_window)

    # Return the boundaries explicitly labeled as Inner/Outer based on the plot convention (8m inner / 2m outer)
    # The actual side is determined by the math, but we label them based on the desired split.
    # Since 8m offset produced the jagged line (red), and 2m produced the smooth line (blue), 
    # we label b1_df as inner (8m) and b2_df as outer (2m) to match your visual result.
    inner_boundary_df = b1_df  # Max 8m offset (Red line in your image)
    outer_boundary_df = b2_df  # Max 2m offset (Blue line in your image)

    return outer_boundary_df, inner_boundary_df


# --- Configuration and Execution ---

# PARAMETERS
FILENAME = 'gateway-4-telemetry.csv'
LAP_NUMBER = 12
MAX_TRACK_WIDTH_M = 10  # This must be 10m for the 2m/8m split
MIN_OFFSET_RATIO = 0.35  # ADJUSTED: Reduced from 0.45 to 0.35 to aggressively prevent crossover
SMOOTHING_WINDOW = 50    # Maintained at 50 for aggressive smoothing

try:
    # 1. Load Data
    df_telemetry = pd.read_csv(FILENAME)
    print(f"Successfully loaded '{FILENAME}'.")
    
    # 2. Filter Centerline Data
    df_centerline = df_telemetry[df_telemetry['lap'] == LAP_NUMBER].copy()
    
    if df_centerline.empty:
        print(f"Error: No data found for lap {LAP_NUMBER} in {FILENAME}. Check your LAP_NUMBER.")
    else:
        # 3. Generate Boundaries
        # The function returns (outer_boundary, inner_boundary)
        outer_boundary, inner_boundary = create_non_overlapping_boundaries(
            df_centerline, 
            max_track_width=MAX_TRACK_WIDTH_M, 
            min_offset_ratio=MIN_OFFSET_RATIO,
            smoothing_window=SMOOTHING_WINDOW # Pass the new parameter
        )
    
        # 4. Plotting the Results 
        plt.figure(figsize=(10, 10))
        plt.plot(df_centerline['longitude'], df_centerline['latitude'], 
                 color='green', linewidth=3, label=f'Centerline (Lap {LAP_NUMBER})')
        plt.plot(inner_boundary['longitude'], inner_boundary['latitude'], 
                 color='red', linewidth=2, label='Inner Boundary (Max 8m Offset - Smoothed)')
        plt.plot(outer_boundary['longitude'], outer_boundary['latitude'], 
                 color='blue', linewidth=2, label='Outer Boundary (Max 2m Offset - Smoothed)')
                 
        plt.title(f'Dynamically Offset Track Geometry (8m Inner / 2m Outer Split - Smoothed)')
        plt.xlabel('Longitude')
        plt.ylabel('Latitude')
        plt.gca().set_aspect('equal', adjustable='box') 
        plt.legend()
        plt.grid(True)
        plt.show()

except FileNotFoundError:
    print(f"Error: The file '{FILENAME}' was not found.")
    print("Please ensure the CSV file is in the same directory as this script.")
except Exception as e:
    print(f"An unexpected error occurred during processing: {e}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def create_track_boundaries(df, num_slices=1000, smoothing_window=100):
    """
    Analyzes telemetry data to determine track boundaries by slicing time.

    Args:
        df (pd.DataFrame): DataFrame with 'latitude', 'longitude', and 'lap_seconds_elapsed'.
        num_slices (int): Number of slices to divide each lap into.
        smoothing_window (int): Window size for smoothing the boundary lines.

    Returns:
        tuple: A tuple containing two DataFrames for the left and right boundaries.
    """
    # Ensure the dataframe is sorted by time
    df['race_seconds_elapsed'] = pd.to_numeric(df['race_seconds_elapsed'])
    df = df.sort_values(by='race_seconds_elapsed').reset_index(drop=True)

    # Get the total number of laps
    num_laps = df['lap'].max()
    
    # Create an empty DataFrame to hold all boundary points
    all_left_boundary_points = pd.DataFrame()
    all_right_boundary_points = pd.DataFrame()

    for lap in range(1, int(num_laps) + 1):
        lap_df = df[df['lap'] == lap].copy()
        
        if lap_df.empty:
            continue

        lap_duration = lap_df['lap_seconds_elapsed'].max()
        if lap_duration == 0:
            continue
            
        # Create slices based on lap_seconds_elapsed
        slice_duration = lap_duration / num_slices
        slices = np.arange(0, lap_duration, slice_duration)
        
        lap_left_boundary = []
        lap_right_boundary = []

        for i in range(len(slices)):
            start_time = slices[i]
            end_time = slices[i] + slice_duration
            
            slice_data = lap_df[(lap_df['lap_seconds_elapsed'] >= start_time) & (lap_df['lap_seconds_elapsed'] < end_time)]
            
            if not slice_data.empty:
                # Find the centroid of the slice to determine its approximate position
                centroid_lon = slice_data['longitude'].mean()
                
                # Sort points in the slice by their distance from the centroid's longitude
                slice_data['lon_dist_from_center'] = np.abs(slice_data['longitude'] - centroid_lon)
                slice_data = slice_data.sort_values(by='lon_dist_from_center')
                
                # Take the outermost points as boundaries. This is a heuristic that works for most tracks.
                left_point = slice_data.iloc[0]
                right_point = slice_data.iloc[-1]
                
                lap_left_boundary.append({'latitude': left_point['latitude'], 'longitude': left_point['longitude']})
                lap_right_boundary.append({'latitude': right_point['latitude'], 'longitude': right_point['longitude']})

        all_left_boundary_points = pd.concat([all_left_boundary_points, pd.DataFrame(lap_left_boundary)])
        all_right_boundary_points = pd.concat([all_right_boundary_points, pd.DataFrame(lap_right_boundary)])

    # Average the boundary points from all laps for a smoother, more accurate representation
    # This requires a more complex method than simple concatenation.
    # We will instead create a single dataframe and smooth it.
    
    combined_points = pd.concat([all_left_boundary_points, all_right_boundary_points])
    combined_points = combined_points.sort_values(by=['latitude', 'longitude'])

    # Using rolling average on the entire combined set for simplicity,
    # though a more sophisticated method would be to align points by track progression.
    left_boundary_df = combined_points.groupby(np.arange(len(combined_points)) // 2).min().reset_index(drop=True)
    right_boundary_df = combined_points.groupby(np.arange(len(combined_points)) // 2).max().reset_index(drop=True)

    # Smooth the final boundaries
    left_boundary_df['latitude'] = left_boundary_df['latitude'].rolling(window=smoothing_window, center=True, min_periods=1).mean()
    left_boundary_df['longitude'] = left_boundary_df['longitude'].rolling(window=smoothing_window, center=True, min_periods=1).mean()
    right_boundary_df['latitude'] = right_boundary_df['latitude'].rolling(window=smoothing_window, center=True, min_periods=1).mean()
    right_boundary_df['longitude'] = right_boundary_df['longitude'].rolling(window=smoothing_window, center=True, min_periods=1).mean()

    return left_boundary_df, right_boundary_df

# Load the combined telemetry data
try:
    df_telemetry = pd.read_csv('combined_telemetry.csv')
    print("Successfully loaded 'combined_telemetry.csv'")
    
    # Run the function to create boundaries
    left_boundary_df, right_boundary_df = create_track_boundaries(df_telemetry, smoothing_window=100)

    # Plot the results
    plt.figure(figsize=(10, 10))
    plt.scatter(df_telemetry['longitude'], df_telemetry['latitude'], s=1, alpha=0.1, label='All Laps Telemetry')
    plt.plot(left_boundary_df['longitude'], left_boundary_df['latitude'], color='red', linewidth=2, label='Left Boundary')
    plt.plot(right_boundary_df['longitude'], right_boundary_df['latitude'], color='blue', linewidth=2, label='Right Boundary')
    plt.title('Track Boundaries from All Telemetry Data')
    plt.xlabel('Longitude')
    plt.ylabel('Latitude')
    plt.legend()
    plt.grid(True)
    plt.show()

except FileNotFoundError:
    print("Error: 'combined_telemetry.csv' was not found.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
# --- Jupyter Notebook Script: Dynamic Offset Track Generator ---
#
# INSTRUCTIONS:
# 1. Ensure 'gateway-4-telemetry.csv' is in the same directory as this notebook.
# 2. Run this cell.
# 3. Adjust the 'MAX_TRACK_WIDTH_M' and 'MIN_OFFSET_RATIO' parameters below to fine-tune the result.
# 4. The processed data will be saved to CSV files in the current directory, with color info in filenames.
#

# Load necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def smooth_circular(df, window_size):
    """Applies a circular rolling mean to latitude and longitude columns."""
    if len(df) <= window_size:
        return df # Cannot smooth if data size is too small

    # Create a cyclic series by concatenating the start/end segments
    pre_pad = df.iloc[-window_size//2:]
    post_pad = df.iloc[:window_size//2]
    
    df_padded = pd.concat([pre_pad, df, post_pad], ignore_index=True)
    
    # Apply rolling mean to the padded series
    df_padded['latitude'] = df_padded['latitude'].rolling(window=window_size, center=True).mean()
    df_padded['longitude'] = df_padded['longitude'].rolling(window=window_size, center=True).mean()
    
    # Slice back to the original length
    start_index = window_size // 2
    end_index = start_index + len(df)
    
    return df_padded.iloc[start_index:end_index].reset_index(drop=True)


def create_non_overlapping_boundaries(df_centerline, max_track_width=10, min_offset_ratio=0.7, smoothing_window=50):
    """
    Creates parallel track boundaries that avoid overlapping in tight corners
    by dynamically adjusting the offset based on local track curvature.
    
    NOTE: This version uses the configuration that yielded a good result:
    INNER_RATIO = 0.8 (Inner Boundary gets max 8m offset, used for the smoother line)
    OUTSIDE_RATIO = 0.2 (Outer Boundary gets max 2m offset, used for the noisier line)
    The plotting labels have been corrected to match the requested 2m/8m split.

    Args:
        df_centerline (pd.DataFrame): DataFrame with 'latitude' and 'longitude' of the centerline.
        max_track_width (int): The maximum desired width of the track in meters. (Must be 10m for split).
        min_offset_ratio (float): The minimum ratio of the offset at the tightest corner
                                  (e.g., 0.7 means the width can be reduced to 70% of max).
        smoothing_window (int): Size of the rolling mean window to apply to the final boundary coordinates.

    Returns:
        tuple: A tuple containing two DataFrames: 
               (outer_boundary_df, inner_boundary_df) with latitude/longitude.
    """
    if df_centerline.empty:
        raise ValueError("Centerline data is empty. Cannot generate boundaries.")

    # Define asymmetric offset ratios based on the user's successful switch:
    # We assume 'Inner Boundary' in the plot (red line in your image) corresponds to 8m, 
    # and 'Outer Boundary' (blue line in your image) corresponds to 2m.
    # The actual geometric side (positive/negative normal) determines which line is which.
    INNER_8M_RATIO = 0.4 
    OUTER_2M_RATIO = 0.6 
    
    # 1. Coordinate Conversion (Lat/Lon to Local Meters)
    start_lat = df_centerline.iloc[0]['latitude']
    start_lon = df_centerline.iloc[0]['longitude']
    
    LAT_M_PER_DEGREE = 111320.0
    
    # Conversion factor for longitude changes based on latitude (it's a Series)
    LON_M_PER_DEGREE_SERIES = LAT_M_PER_DEGREE * np.cos(np.radians(df_centerline['latitude']))

    # Store the single conversion factor at the START point for consistent reverse projection
    LON_M_PER_DEGREE_START = LON_M_PER_DEGREE_SERIES.iloc[0] 

    df_centerline['x_m'] = (df_centerline['longitude'] - start_lon) * LON_M_PER_DEGREE_SERIES
    df_centerline['y_m'] = (df_centerline['latitude'] - start_lat) * LAT_M_PER_DEGREE

    # 2. Curvature Calculation
    # Calculate first and second derivatives for curvature formula
    dx = np.gradient(df_centerline['x_m'])
    dy = np.gradient(df_centerline['y_m'])
    d2x = np.gradient(dx)
    d2y = np.gradient(dy)
    
    # Curvature formula: k = |x'y'' - y'x''| / (x'^2 + y'^2)^(3/2)
    curvature = np.abs(dx * d2y - dy * d2x) / ((dx**2 + dy**2)**1.5)
    
    # 3. Dynamic Offset Scaling
    if curvature.max() == curvature.min():
        normalized_curvature = np.zeros_like(curvature)
    else:
        normalized_curvature = (curvature - curvature.min()) / (curvature.max() - curvature.min())
    
    # Calculate the scale factor: 1.0 (straight) down to min_offset_ratio (tightest corner)
    offset_scale = (1 - normalized_curvature) * (1.0 - min_offset_ratio) + min_offset_ratio
    
    # Calculate asymmetric dynamic offsets
    dynamic_offset_8m = (max_track_width * INNER_8M_RATIO) * offset_scale
    dynamic_offset_2m = (max_track_width * OUTER_2M_RATIO) * offset_scale

    # 4. Normal Vector Calculation
    magnitude = np.sqrt(dx**2 + dy**2)
    dx_norm = dx / magnitude
    dy_norm = dy / magnitude
    
    # Normal vector is perpendicular to the tangent ([-dy_norm, dx_norm])
    nx, ny = -dy_norm, dx_norm

    # 5. Calculate Offset Points
    # Boundary 1 (Corresponds to 8m offset, mapped to red line in your image)
    b1_x = df_centerline['x_m'] + nx * dynamic_offset_8m
    b1_y = df_centerline['y_m'] + ny * dynamic_offset_8m
    
    # Boundary 2 (Corresponds to 2m offset, mapped to blue line in your image)
    b2_x = df_centerline['x_m'] - nx * dynamic_offset_2m
    b2_y = df_centerline['y_m'] - ny * dynamic_offset_2m
    
    # 6. Convert back to Lat/Lon
    b1_df = pd.DataFrame({
        'latitude': start_lat + b1_y / LAT_M_PER_DEGREE,
        'longitude': start_lon + b1_x / LON_M_PER_DEGREE_START
    })
    
    b2_df = pd.DataFrame({
        'latitude': start_lat + b2_y / LAT_M_PER_DEGREE,
        'longitude': start_lon + b2_x / LON_M_PER_DEGREE_START
    })

    # 7. Post-processing Smoothing (using circular rolling mean)
    if smoothing_window > 1:
        b1_df = smooth_circular(b1_df, smoothing_window)
        b2_df = smooth_circular(b2_df, smoothing_window)

    # Return the boundaries explicitly labeled as Inner/Outer based on the plot convention (8m inner / 2m outer)
    # The actual side is determined by the math, but we label them based on the desired split.
    # Since 8m offset produced the jagged line (red), and 2m produced the smooth line (blue), 
    # we label b1_df as inner (8m) and b2_df as outer (2m) to match your visual result.
    inner_boundary_df = b1_df  # Max 8m offset (Red line in your image)
    outer_boundary_df = b2_df  # Max 2m offset (Blue line in your image)

    return outer_boundary_df, inner_boundary_df


# --- Configuration and Execution ---

# PARAMETERS
FILENAME = 'gateway-2-telemetry.csv'
LAP_NUMBER = 1

MAX_TRACK_WIDTH_M = 10  # This must be 10m for the 2m/8m split
MIN_OFFSET_RATIO = 0.35  # ADJUSTED: Reduced from 0.45 to 0.35 to aggressively prevent crossover
SMOOTHING_WINDOW = 50    # Maintained at 50 for aggressive smoothing

# Define colors for each trace (to use in filenames and plotting)
CENTERLINE_COLOR = 'green'
INNER_BOUNDARY_COLOR = 'red'
OUTER_BOUNDARY_COLOR = 'blue'

OUTPUT_PREFIX = f'gateway_lap{LAP_NUMBER}'  # Base prefix for output filenames

try:
    # 1. Load Data
    df_telemetry = pd.read_csv(FILENAME)
    print(f"Successfully loaded '{FILENAME}'.")
    
    # 2. Filter Centerline Data
    df_centerline = df_telemetry[df_telemetry['lap'] == LAP_NUMBER].copy()
    
    if df_centerline.empty:
        print(f"Error: No data found for lap {LAP_NUMBER} in {FILENAME}. Check your LAP_NUMBER.")
    else:
        # 3. Generate Boundaries
        # The function returns (outer_boundary, inner_boundary)
        outer_boundary, inner_boundary = create_non_overlapping_boundaries(
            df_centerline, 
            max_track_width=MAX_TRACK_WIDTH_M, 
            min_offset_ratio=MIN_OFFSET_RATIO,
            smoothing_window=SMOOTHING_WINDOW # Pass the new parameter
        )
    
        # 4. Save the processed data to CSV files with color information in filenames
        centerline_filename = f"{OUTPUT_PREFIX}_centerline_{CENTERLINE_COLOR}.csv"
        inner_boundary_filename = f"{OUTPUT_PREFIX}_inner_8m_{INNER_BOUNDARY_COLOR}.csv"
        outer_boundary_filename = f"{OUTPUT_PREFIX}_outer_2m_{OUTER_BOUNDARY_COLOR}.csv"
        
        # Save centerline data (select only needed columns)
        df_centerline[['latitude', 'longitude']].to_csv(centerline_filename, index=False)
        print(f"Centerline data saved to '{centerline_filename}'")
        
        # Save boundary data
        inner_boundary.to_csv(inner_boundary_filename, index=False)
        print(f"Inner boundary data (8m offset - {INNER_BOUNDARY_COLOR}) saved to '{inner_boundary_filename}'")
        
        outer_boundary.to_csv(outer_boundary_filename, index=False)
        print(f"Outer boundary data (2m offset - {OUTER_BOUNDARY_COLOR}) saved to '{outer_boundary_filename}'")
        
        # 5. Plotting the Results 
        plt.figure(figsize=(10, 10))
        plt.plot(df_centerline['longitude'], df_centerline['latitude'], 
                 color=CENTERLINE_COLOR, linewidth=3, label=f'Centerline (Lap {LAP_NUMBER})')
        plt.plot(inner_boundary['longitude'], inner_boundary['latitude'], 
                 color=INNER_BOUNDARY_COLOR, linewidth=2, label='Inner Boundary (Max 8m Offset - Smoothed)')
        plt.plot(outer_boundary['longitude'], outer_boundary['latitude'], 
                 color=OUTER_BOUNDARY_COLOR, linewidth=2, label='Outer Boundary (Max 2m Offset - Smoothed)')
                 
        plt.title(f'Dynamically Offset Track Geometry (8m Inner / 2m Outer Split - Smoothed)')
        plt.xlabel('Longitude')
        plt.ylabel('Latitude')
        plt.gca().set_aspect('equal', adjustable='box') 
        plt.legend()
        plt.grid(True)
        plt.show()

except FileNotFoundError:
    print(f"Error: The file '{FILENAME}' was not found.")
    print("Please ensure the CSV file is in the same directory as this script.")
except Exception as e:
    print(f"An unexpected error occurred during processing: {e}")
```

## Simple racing line calculation

```python
import pandas as pd
import numpy as np
from scipy.interpolate import interp1d
import scipy.spatial as spatial
from scipy.signal import savgol_filter
import math

def haversine_distance(lat1, lon1, lat2, lon2):
    """
    Calculate the great circle distance between two points 
    on the earth (specified in decimal degrees)
    """
    # Convert decimal degrees to radians
    lat1, lon1, lat2, lon2 = map(math.radians, [lat1, lon1, lat2, lon2])
    
    # Haversine formula
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = math.sin(dlat/2)**2 + math.cos(lat1) * math.cos(lat2) * math.sin(dlon/2)**2
    c = 2 * math.asin(math.sqrt(a))
    r = 6371000  # Radius of earth in meters
    return c * r

def calculate_distances(lats, lons):
    """Calculate distances between consecutive coordinate points using haversine"""
    distances = []
    for i in range(1, len(lats)):
        dist = haversine_distance(lats[i-1], lons[i-1], lats[i], lons[i])
        distances.append(dist)
    return np.array(distances)

def calculate_ideal_racing_line(df_combined):
    """
    Calculates a simplified ideal racing line by finding the centerline
    of the track and smoothing it.
    """
    # Separate inner and outer traces
    df_inner = df_combined[df_combined['trace_type'] == 'inner'].copy()
    df_outer = df_combined[df_combined['trace_type'] == 'outer'].copy()

    # Ensure we have enough data points
    if len(df_inner) < 10 or len(df_outer) < 10:
        print(f"Warning: Limited data points. Inner: {len(df_inner)}, Outer: {len(df_outer)}")
        
    # Find ordered points along the track (assuming roughly oval/circular track)
    # First, find the centroid of all points to determine track orientation
    centroid_lat = np.mean(df_combined['latitude'])
    centroid_lon = np.mean(df_combined['longitude'])
    
    # Calculate angle from centroid for sorting points along the track
    df_inner['angle'] = np.arctan2(
        df_inner['longitude'] - centroid_lon, 
        df_inner['latitude'] - centroid_lat
    )
    df_outer['angle'] = np.arctan2(
        df_outer['longitude'] - centroid_lon, 
        df_outer['latitude'] - centroid_lat
    )
    
    # Sort by angle to follow the track
    df_inner = df_inner.sort_values('angle').reset_index(drop=True)
    df_outer = df_outer.sort_values('angle').reset_index(drop=True)
    
    # Use the shorter of the two traces to avoid extrapolation issues
    min_length = min(len(df_inner), len(df_outer))
    
    # Resample the longer trace to match the shorter one
    if len(df_inner) > min_length:
        indices = np.linspace(0, len(df_inner)-1, min_length).astype(int)
        df_inner = df_inner.iloc[indices].reset_index(drop=True)
    elif len(df_outer) > min_length:
        indices = np.linspace(0, len(df_outer)-1, min_length).astype(int)
        df_outer = df_outer.iloc[indices].reset_index(drop=True)
    
    # Calculate centerline as midpoint between inner and outer
    df_centerline = pd.DataFrame({
        'latitude': (df_inner['latitude'].values + df_outer['latitude'].values) / 2,
        'longitude': (df_inner['longitude'].values + df_outer['longitude'].values) / 2
    })
    
    # Apply smoothing if enough data points
    window_length = min(51, len(df_centerline) - 2)
    if window_length % 2 == 0:
        window_length -= 1  # Must be odd
    
    if window_length > 2:
        polyorder = min(3, window_length - 1)
        df_centerline['lat_smooth'] = savgol_filter(df_centerline['latitude'], window_length, polyorder)
        df_centerline['lon_smooth'] = savgol_filter(df_centerline['longitude'], window_length, polyorder)
    else:
        # If not enough points for smoothing, use original values
        df_centerline['lat_smooth'] = df_centerline['latitude']
        df_centerline['lon_smooth'] = df_centerline['longitude']
    
    return df_centerline

def calculate_curvature(lats, lons, distances):
    """
    Calculate the curvature using cumulative distance along the path for more accurate results
    """
    # Create cumulative distance array for parameterization
    cum_distances = np.insert(np.cumsum(distances), 0, 0)
    
    # Fit spline to get smoother derivatives
    if len(cum_distances) > 3:
        try:
            lat_spline = interp1d(cum_distances, lats, kind='cubic')
            lon_spline = interp1d(cum_distances, lons, kind='cubic')
            
            # Evaluate at regular intervals
            t = np.linspace(cum_distances[0], cum_distances[-1], len(cum_distances) * 2)
            
            # Get coordinates and derivatives
            lat_vals = lat_spline(t)
            lon_vals = lon_spline(t)
            
            # Calculate first derivatives with respect to distance
            lat_prime = np.gradient(lat_vals, t)
            lon_prime = np.gradient(lon_vals, t)
            
            # Calculate second derivatives
            lat_prime2 = np.gradient(lat_prime, t)
            lon_prime2 = np.gradient(lon_prime, t)
            
            # Curvature formula for parametric curve
            numerator = np.abs(lat_prime * lon_prime2 - lon_prime * lat_prime2)
            denominator = (lat_prime**2 + lon_prime**2)**1.5
            
            # Avoid division by zero
            denominator[denominator == 0] = np.finfo(float).eps
            curvature = numerator / denominator
            
            # Interpolate back to original points
            curve_interp = interp1d(t, curvature, bounds_error=False, fill_value='extrapolate')
            result = curve_interp(cum_distances)
            
            return result
        except:
            print("Error in curvature calculation, using fallback method")
            return np.ones(len(lats)) * 0.01  # Fallback constant curvature
    else:
        # Not enough points for spline, return constant
        return np.ones(len(lats)) * 0.01

def estimate_speed(curvature, max_speed_mph=55, min_speed_mph=15):
    """
    Estimate speed based on curvature
    Higher curvature = lower speed, Lower curvature = higher speed
    """
    max_speed = max_speed_mph * 0.44704  # Convert mph to m/s
    min_speed = min_speed_mph * 0.44704  # Convert mph to m/s
    
    # Add a small constant to avoid division by zero
    curvature = curvature + 1e-10
    
    # Racing line theory: v^2 = a/r where a is lateral acceleration, r is radius (1/curvature)
    # Using a simple model: speed = sqrt(constant / curvature)
    max_curvature = np.percentile(curvature, 95)  # 95th percentile to avoid outliers
    
    # Scale curvature to get appropriate speeds
    scaled_curvature = curvature / max_curvature
    
    # Calculate speed: max_speed at straight sections (low curvature) 
    # and min_speed at max curvature
    raw_speed = max_speed - (max_speed - min_speed) * np.minimum(scaled_curvature, 1.0)
    
    # Apply smoothing to speed transitions if enough points
    if len(raw_speed) > 5:
        window_length = min(51, len(raw_speed) - 2)
        if window_length % 2 == 0:
            window_length -= 1  # Must be odd
        polyorder = min(3, window_length - 1)
        speed = savgol_filter(raw_speed, window_length, polyorder)
    else:
        speed = raw_speed
    
    # Convert back to mph for output
    speed_mph = speed / 0.44704
    
    return speed_mph

def resample_line(df_line, frequency_hz=100):
    """
    Resamples a line at a specific frequency to create a dense trace with speed estimates.
    """
    # Calculate real distances between points using haversine formula
    distances = calculate_distances(df_line['lat_smooth'].values, df_line['lon_smooth'].values)
    
    # Calculate curvature for speed estimation
    curvature = calculate_curvature(
        df_line['lat_smooth'].values, 
        df_line['lon_smooth'].values, 
        distances
    )
    
    # Estimate speed based on curvature
    speeds = estimate_speed(curvature, max_speed_mph=55, min_speed_mph=15)
    
    # Create a cumulative distance array for interpolation
    cumulative_dist = np.insert(np.cumsum(distances), 0, 0)
    
    # Estimate total time based on distances and speeds
    # For each segment, time = distance/speed
    segment_times = distances / speeds[:-1]  # speeds[:-1] to match the length of distances
    cumulative_time = np.insert(np.cumsum(segment_times), 0, 0)
    total_time = cumulative_time[-1]
    
    print(f"Track length: {cumulative_dist[-1]:.2f} meters")
    print(f"Estimated lap time: {total_time:.2f} seconds")
    
    # Create interpolation functions
    f_lat = interp1d(cumulative_time, df_line['lat_smooth'], bounds_error=False, fill_value='extrapolate')
    f_lon = interp1d(cumulative_time, df_line['lon_smooth'], bounds_error=False, fill_value='extrapolate')
    f_dist = interp1d(cumulative_time, cumulative_dist, bounds_error=False, fill_value='extrapolate')
    f_speed = interp1d(cumulative_time, speeds, bounds_error=False, fill_value='extrapolate')
    
    # Generate points at the desired frequency
    num_points = int(total_time * frequency_hz) + 1
    if num_points < 10:  # Ensure we have at least 10 points
        num_points = 10
        print(f"Warning: Increasing points to minimum of {num_points}")
        
    new_times = np.linspace(0, total_time, num_points)
    
    # Generate the new coordinates, distances and speeds
    resampled_df = pd.DataFrame({
        'latitude': f_lat(new_times),
        'longitude': f_lon(new_times),
        'distance': f_dist(new_times),
        'speed_mph': f_speed(new_times),
        'time': new_times
    })
    
    # Add a unique ID for visualization in Kepler.gl
    resampled_df['id'] = np.arange(len(resampled_df))
    
    return resampled_df

# Main script execution
if __name__ == '__main__':
    # Load your combined track data
    combined_file = 'track.csv'
    df_combined = pd.read_csv(combined_file)
    
    print(f"Loaded track with {len(df_combined)} data points")
    print(f"Inner points: {len(df_combined[df_combined['trace_type'] == 'inner'])}")
    print(f"Outer points: {len(df_combined[df_combined['trace_type'] == 'outer'])}")
    
    # Calculate the ideal racing line
    df_racing_line = calculate_ideal_racing_line(df_combined)
    print(f"Generated centerline with {len(df_racing_line)} points")
    
    # Resample the line to 100 Hz with speed estimates
    df_resampled_line = resample_line(df_racing_line, frequency_hz=100)
    print(f"Resampled to {len(df_resampled_line)} points at 100Hz")
    
    # Save the resampled line for visualization
    output_file = 'gateway_racing_line_100hz.csv'
    df_resampled_line.to_csv(output_file, index=False)
    
    print(f"Ideal racing line calculated and saved to {output_file}")
    print(f"Speed range: {df_resampled_line['speed_mph'].min():.1f} - {df_resampled_line['speed_mph'].max():.1f} mph")
```

```python
import pandas as pd
import numpy as np
import scipy.spatial as spatial
from scipy.interpolate import interp1d

def analyze_and_project_lap(telemetry_file, ideal_line_file, race_number, lap_number):
    """
    Reads telemetry data and projects it onto a calculated ideal racing line.

    Args:
        telemetry_file (str): Path to the driver's telemetry CSV.
        ideal_line_file (str): Path to the ideal racing line CSV.
        race_number (int): The specific race number to analyze.
        lap_number (int): The specific lap number to analyze.
    
    Returns:
        pandas.DataFrame: A DataFrame with telemetry data projected onto the ideal line.
    """
    try:
        # Load telemetry and ideal racing line data
        df_telemetry = pd.read_csv(telemetry_file)
        df_ideal = pd.read_csv(ideal_line_file)

        # Filter the telemetry data for the specified race and lap
        df_lap = df_telemetry[(df_telemetry['race'] == race_number) & (df_telemetry['lap'] == lap_number)].copy()
        
        if df_lap.empty:
            print(f"Error: No data found for Race {race_number}, Lap {lap_number}.")
            return None

        # Sort ideal line points by their ID to ensure sequential order
        df_ideal = df_ideal.sort_values('id').reset_index(drop=True)

        # Create a K-D tree for the ideal racing line's coordinates
        tree = spatial.cKDTree(df_ideal[['X', 'Y']])

        # For each point in the driver's lap, find the closest point on the ideal line
        distances, indices = tree.query(df_lap[['latitude', 'longitude']])

        # Map driver's telemetry data to the ideal line's coordinates
        # We'll create a new DataFrame to store the projected data
        df_projected = df_ideal.iloc[indices].copy().reset_index(drop=True)

        # Add the driver's original telemetry data to the new DataFrame
        # This aligns the telemetry values with the ideal line's coordinates
        telemetry_columns = ['time', 'speed', 'speed_mph', 'accel_longitudinal', 'accel_lateral', 'g_force_smoothed']
        for col in telemetry_columns:
            df_projected[f'driver_{col}'] = df_lap[col].values

        # Calculate deviation from the ideal line
        # This shows how far off the racing line the driver was at any given point
        df_projected['deviation_from_ideal'] = distances

        print(f"Analysis complete. Telemetry from Race {race_number}, Lap {lap_number} projected onto the ideal racing line.")
        
        return df_projected

    except FileNotFoundError as e:
        print(f"Error: The file '{e.filename}' was not found. Please check your file paths.")
        return None
    except KeyError as e:
        print(f"Error: Missing column {e}. Ensure all CSV files have the required headers.")
        return None

# --- Example Usage ---
if __name__ == '__main__':
    telemetry_file_path = 'gateway-1-telemetry.csv'
    ideal_line_file_path = 'gateway_racing_line_100hz.csv'

    # Specify the race and lap to analyze
    target_race = 1
    target_lap = 7

    # Run the analysis and get the projected data
    projected_data = analyze_and_project_lap(telemetry_file_path, ideal_line_file_path, target_race, target_lap)

    if projected_data is not None:
        # Save the result to a new CSV for visualization and further analysis
        output_file = 'gateway_projected_analysis.csv'
        projected_data.to_csv(output_file, index=False)
        print(f"\nProjected data saved to {output_file}.")
        print("This file can be used to visualize the driver's performance on the ideal racing line.")
```

```python
import pandas as pd
import numpy as np
from scipy.interpolate import interp1d
import scipy.spatial as spatial
from scipy.signal import savgol_filter
import math
from geopy.distance import geodesic

def haversine_distance(lat1, lon1, lat2, lon2):
    """Calculate the great circle distance between two points in meters"""
    return geodesic((lat1, lon1), (lat2, lon2)).meters

def bearing(lat1, lon1, lat2, lon2):
    """Calculate the bearing between two points in degrees"""
    lat1, lon1, lat2, lon2 = map(math.radians, [lat1, lon1, lat2, lon2])
    dlon = lon2 - lon1
    y = math.sin(dlon) * math.cos(lat2)
    x = math.cos(lat1) * math.sin(lat2) - math.sin(lat1) * math.cos(lat2) * math.cos(dlon)
    bearing_rad = math.atan2(y, x)
    bearing_deg = (math.degrees(bearing_rad) + 360) % 360
    return bearing_deg

def calculate_track_width(df_inner, df_outer, df_center):
    """Calculate track width at each centerline point"""
    # Build KD trees for inner and outer boundaries
    inner_tree = spatial.cKDTree(df_inner[['latitude', 'longitude']].values)
    outer_tree = spatial.cKDTree(df_outer[['latitude', 'longitude']].values)
    
    # For each centerline point, find the nearest inner and outer points
    center_points = df_center[['latitude', 'longitude']].values
    
    dist_to_inner, idx_inner = inner_tree.query(center_points)
    dist_to_outer, idx_outer = outer_tree.query(center_points)
    
    inner_points = df_inner.iloc[idx_inner][['latitude', 'longitude']].values
    outer_points = df_outer.iloc[idx_outer][['latitude', 'longitude']].values
    
    # Calculate track width at each point
    track_widths = np.array([
        haversine_distance(inner_points[i][0], inner_points[i][1], outer_points[i][0], outer_points[i][1])
        for i in range(len(center_points))
    ])
    
    return track_widths

def calculate_curvature(latitudes, longitudes):
    """
    Calculate the curvature of a track at each point.
    A more robust method that handles GPS coordinates better.
    """
    n = len(latitudes)
    
    # Need at least 3 points to compute curvature
    if n < 3:
        return np.zeros(n)
    
    # Calculate the distance between consecutive points
    distances = []
    for i in range(1, n):
        d = haversine_distance(latitudes[i-1], longitudes[i-1], latitudes[i], longitudes[i])
        distances.append(d)
    
    # Calculate cumulative distance
    cum_dist = np.insert(np.cumsum(distances), 0, 0)
    
    # Approximate the curvature using finite differences
    curvature = np.zeros(n)
    
    for i in range(1, n-1):
        # Get the three consecutive points
        if i > 0 and i < n-1:
            # Convert to local x-y coordinates
            p1 = (0, 0)
            
            d2 = haversine_distance(latitudes[i-1], longitudes[i-1], latitudes[i], longitudes[i])
            b2 = bearing(latitudes[i-1], longitudes[i-1], latitudes[i], longitudes[i])
            p2 = (d2 * math.sin(math.radians(b2)), d2 * math.cos(math.radians(b2)))
            
            d3 = haversine_distance(latitudes[i-1], longitudes[i-1], latitudes[i+1], longitudes[i+1])
            b3 = bearing(latitudes[i-1], longitudes[i-1], latitudes[i+1], longitudes[i+1])
            p3 = (d3 * math.sin(math.radians(b3)), d3 * math.cos(math.radians(b3)))
            
            # Circumcircle calculation
            D = 2 * (p1[0] * (p2[1] - p3[1]) + p2[0] * (p3[1] - p1[1]) + p3[0] * (p1[1] - p2[1]))
            
            if abs(D) > 1e-10:  # Avoid division by near-zero
                ux = ((p1[0]**2 + p1[1]**2) * (p2[1] - p3[1]) + 
                      (p2[0]**2 + p2[1]**2) * (p3[1] - p1[1]) + 
                      (p3[0]**2 + p3[1]**2) * (p1[1] - p2[1])) / D
                
                uy = ((p1[0]**2 + p1[1]**2) * (p3[0] - p2[0]) + 
                      (p2[0]**2 + p2[1]**2) * (p1[0] - p3[0]) + 
                      (p3[0]**2 + p3[1]**2) * (p2[0] - p1[0])) / D
                
                # Radius of the circle
                r = math.sqrt((p1[0] - ux)**2 + (p1[1] - uy)**2)
                
                # Curvature = 1/r
                curvature[i] = 1.0 / max(r, 1e-10)  # Avoid division by zero
            else:
                curvature[i] = 0.0  # Straight line segment
    
    # For the endpoints, use the same curvature as their neighbors
    curvature[0] = curvature[1]
    curvature[n-1] = curvature[n-2]
    
    # Smooth the curvature values
    window_size = min(11, n - 2)
    if window_size % 2 == 0:
        window_size -= 1  # Must be odd
    
    if window_size >= 3:
        curvature = savgol_filter(curvature, window_size, 2)
    
    return curvature

def optimize_racing_line(df_center, df_inner, df_outer, smoothing=True):
    """
    Optimize the racing line by shifting centerline points towards the inside of turns
    with improved smoothing for a more natural racing trajectory.
    """
    # Calculate track width at each point
    track_widths = calculate_track_width(df_inner, df_outer, df_center)
    
    # Pre-smooth the centerline before calculating curvature
    window_size = min(51, len(df_center) - 2)
    if window_size % 2 == 0:
        window_size -= 1  # Must be odd
    
    if window_size >= 3:
        smoothed_lats = savgol_filter(df_center['latitude'].values, window_size, 3)
        smoothed_lons = savgol_filter(df_center['longitude'].values, window_size, 3)
    else:
        smoothed_lats = df_center['latitude'].values
        smoothed_lons = df_center['longitude'].values
    
    # Calculate curvature of the smoothed centerline
    curvature = calculate_curvature(smoothed_lats, smoothed_lons)
    
    # Apply additional smoothing to the curvature values
    if len(curvature) > 10:
        curvature = savgol_filter(curvature, min(51, len(curvature) - 2), 3)
    
    # Calculate direction vectors (bearing) between consecutive points
    bearings = []
    for i in range(len(smoothed_lats)-1):
        bearings.append(bearing(smoothed_lats[i], smoothed_lons[i], 
                               smoothed_lats[i+1], smoothed_lons[i+1]))
    
    # Add the last bearing (use same as the previous point)
    bearings.append(bearings[-1])
    bearings = np.array(bearings)
    
    # Smooth the bearings to avoid sudden direction changes
    if len(bearings) > 10:
        bearings = savgol_filter(bearings, min(51, len(bearings) - 2), 3)
    
    # Calculate normal vectors (90 degrees to direction) - right side normal
    normal_bearings = (bearings + 90) % 360
    
    # Determine turn direction using a more robust method:
    # Calculate change in bearing over a window rather than just consecutive points
    window = 5
    bearing_changes = []
    
    for i in range(len(bearings)):
        # Get indices for the window, handling wrap-around for circular track
        indices = [(i + j) % len(bearings) for j in range(-window//2, window//2 + 1)]
        window_bearings = bearings[indices]
        
        # Calculate bearing change (handling the 0-360 discontinuity)
        changes = np.diff(window_bearings)
        changes = np.where(changes > 180, changes - 360, changes)
        changes = np.where(changes < -180, changes + 360, changes)
        
        # Average change indicates turn direction
        bearing_changes.append(np.mean(changes))
    
    turn_direction = np.sign(bearing_changes)
    
    # Smooth the turn direction
    if len(turn_direction) > 10:
        # Use a larger window for turn direction smoothing
        turn_window = min(101, len(turn_direction) - 2)
        if turn_window % 2 == 0:
            turn_window -= 1
        turn_direction = savgol_filter(turn_direction, turn_window, 1)
    
    # Calculate deviation amount (how far from center to move)
    # Use a more conservative approach - maximum 30% of track width
    max_deviation_pct = 0.3
    
    # Normalize curvature more robustly using percentiles
    normalized_curvature = np.zeros_like(curvature)
    if np.max(curvature) > 0:
        # Use 95th percentile for more stable normalization
        p95_curvature = np.percentile(curvature, 95)
        normalized_curvature = curvature / p95_curvature
        normalized_curvature = np.clip(normalized_curvature, 0, 1)
    
    # Apply smoothing to normalized curvature
    if len(normalized_curvature) > 10:
        normalized_curvature = savgol_filter(normalized_curvature, min(51, len(normalized_curvature) - 2), 3)
    
    # Calculate the deviation percentage with a quadratic effect
    # This makes the racing line more aggressive at high curvature areas
    deviation_pct = normalized_curvature**2 * max_deviation_pct
    
    # Apply turn direction to deviation
    signed_deviation = deviation_pct * -turn_direction
    
    # Smooth the final deviation values
    if len(signed_deviation) > 10:
        signed_deviation = savgol_filter(signed_deviation, min(51, len(signed_deviation) - 2), 3)
    
    # Calculate the actual deviation distance in meters
    deviation_distance = track_widths * signed_deviation
    
    # Apply the deviation to get new points
    new_latitudes = []
    new_longitudes = []
    
    for i in range(len(smoothed_lats)):
        # Convert deviation distance and bearing to latitude/longitude offset
        lat = smoothed_lats[i]
        lon = smoothed_lons[i]
        
        # Earth radius in meters
        earth_radius = 6371000
        
        # Convert deviation distance to degrees
        dist_deg_lat = math.degrees(deviation_distance[i] / earth_radius)
        dist_deg_lon = math.degrees(deviation_distance[i] / (earth_radius * math.cos(math.radians(lat))))
        
        # Calculate offsets based on normal bearing
        normal_rad = math.radians(normal_bearings[i])
        lat_offset = dist_deg_lat * math.cos(normal_rad)
        lon_offset = dist_deg_lon * math.sin(normal_rad)
        
        # Apply offsets
        new_latitudes.append(lat + lat_offset)
        new_longitudes.append(lon + lon_offset)
    
    # Create DataFrame for the optimized racing line
    df_racing_line = pd.DataFrame({
        'latitude': new_latitudes,
        'longitude': new_longitudes,
        'curvature': curvature,
        'track_width': track_widths,
        'deviation_pct': signed_deviation,
        'bearing': bearings
    })
    
    # Apply final smoothing to the racing line if requested
    if smoothing:
        # Use a larger window for final smoothing to get a very smooth line
        window_size = min(101, len(df_racing_line) - 2)
        if window_size % 2 == 0:
            window_size -= 1  # Must be odd
        
        if window_size >= 3:
            df_racing_line['latitude'] = savgol_filter(df_racing_line['latitude'], window_size, 3)
            df_racing_line['longitude'] = savgol_filter(df_racing_line['longitude'], window_size, 3)
    
    return df_racing_line

def calculate_speeds(df_racing_line, max_speed_mph=55, min_speed_mph=15):
    """
    Calculate speeds based on curvature of the racing line.
    Using the formula: speed^2 = constant / curvature.
    """
    # Convert speeds to m/s for calculations
    max_speed = max_speed_mph * 0.44704
    min_speed = min_speed_mph * 0.44704
    
    # Get the curvature values
    curvature = df_racing_line['curvature'].values
    
    # Add a small value to avoid division by zero
    curvature = np.maximum(curvature, 1e-10)
    
    # Determine a constant that gives our max speed on the straightest section
    min_curvature = np.percentile(curvature, 10)  # 10th percentile for a robust minimum
    constant = max_speed**2 * min_curvature
    
    # Calculate raw speeds using the racing formula v^2 = constant/curvature
    raw_speeds_ms = np.sqrt(constant / curvature)
    
    # Clamp speeds between min and max
    speeds_ms = np.clip(raw_speeds_ms, min_speed, max_speed)
    
    # Smooth the speed transitions
    window_size = min(11, len(speeds_ms) - 2)
    if window_size % 2 == 0:
        window_size -= 1
    
    if window_size >= 3:
        speeds_ms = savgol_filter(speeds_ms, window_size, 2)
    
    # Convert back to mph for output
    speeds_mph = speeds_ms / 0.44704
    
    return speeds_mph

def finalize_racing_line(df_racing_line, frequency_hz=100):
    """
    Calculate the final racing line with speeds and resample at the desired frequency.
    """
    # Calculate the speed for each point
    speeds_mph = calculate_speeds(df_racing_line, max_speed_mph=55, min_speed_mph=15)
    df_racing_line['speed_mph'] = speeds_mph
    
    # Calculate distances between consecutive points
    lats = df_racing_line['latitude'].values
    lons = df_racing_line['longitude'].values
    
    distances = []
    for i in range(1, len(lats)):
        d = haversine_distance(lats[i-1], lons[i-1], lats[i], lons[i])
        distances.append(d)
    
    # Calculate cumulative distance
    cumulative_dist = np.insert(np.cumsum(distances), 0, 0)
    df_racing_line['distance'] = cumulative_dist
    
    # Calculate time based on speed and distance
    speeds_mps = df_racing_line['speed_mph'].values * 0.44704  # Convert mph to m/s
    
    # Calculate time for each segment (using average speed between consecutive points)
    times = [0]  # Start with 0 for the first point
    
    for i in range(len(distances)):
        # Use average speed over segment
        if i < len(speeds_mps) - 1:
            avg_speed = (speeds_mps[i] + speeds_mps[i+1]) / 2
            # Avoid division by zero
            if avg_speed > 0:
                time = distances[i] / avg_speed
            else:
                time = 0
            times.append(times[-1] + time)
    
    # Ensure times array matches the length of the DataFrame
    if len(times) < len(df_racing_line):
        # Add last time again if needed
        times.append(times[-1])
    elif len(times) > len(df_racing_line):
        # Trim if too long
        times = times[:len(df_racing_line)]
    
    df_racing_line['time'] = times
    
    # Resample to the desired frequency if requested
    if frequency_hz > 0:
        total_time = times[-1]
        num_points = int(total_time * frequency_hz) + 1
        
        new_times = np.linspace(0, total_time, num_points)
        
        # Create interpolation functions
        f_lat = interp1d(times, df_racing_line['latitude'], bounds_error=False, fill_value='extrapolate')
        f_lon = interp1d(times, df_racing_line['longitude'], bounds_error=False, fill_value='extrapolate')
        f_speed = interp1d(times, df_racing_line['speed_mph'], bounds_error=False, fill_value='extrapolate')
        f_dist = interp1d(times, df_racing_line['distance'], bounds_error=False, fill_value='extrapolate')
        f_bear = interp1d(times, df_racing_line['bearing'], bounds_error=False, fill_value='extrapolate')
        
        # Generate the resampled points
        resampled_df = pd.DataFrame({
            'latitude': f_lat(new_times),
            'longitude': f_lon(new_times),
            'speed_mph': f_speed(new_times),
            'distance': f_dist(new_times),
            'bearing': f_bear(new_times),
            'time': new_times
        })
        
        # Add ID column for visualization
        resampled_df['id'] = np.arange(len(resampled_df))
        
        return resampled_df
    else:
        # Add ID column for the original dataframe
        df_racing_line['id'] = np.arange(len(df_racing_line))
        return df_racing_line

# Main script execution
if __name__ == '__main__':
    # Load your centerline and track boundary data
    centerline_file = 'centerline.csv'
    track_file = 'track.csv'
    
    df_center = pd.read_csv(centerline_file)
    df_track = pd.read_csv(track_file)
    
    # Separate inner and outer boundaries
    df_inner = df_track[df_track['trace_type'] == 'inner'].copy()
    df_outer = df_track[df_track['trace_type'] == 'outer'].copy()
    
    print(f"Loaded centerline with {len(df_center)} points")
    print(f"Loaded track with {len(df_inner)} inner points and {len(df_outer)} outer points")
    
    # Optimize the racing line based on the centerline and track boundaries
    df_racing_line = optimize_racing_line(df_center, df_inner, df_outer, smoothing=True)
    print(f"Optimized racing line with {len(df_racing_line)} points")
    
    # Finalize the racing line with speeds and resample to 100 Hz
    df_resampled = finalize_racing_line(df_racing_line, frequency_hz=100)
    print(f"Resampled racing line to {len(df_resampled)} points at 100Hz")
    
    # Save the resampled racing line for visualization
    output_file = 'racing_line.csv'
    df_resampled.to_csv(output_file, index=False)
    
    print(f"Racing line calculated and saved to {output_file}")
    print(f"Speed range: {df_resampled['speed_mph'].min():.1f} - {df_resampled['speed_mph'].max():.1f} mph")
    print(f"Total distance: {df_resampled['distance'].max():.1f} meters")
    print(f"Total time: {df_resampled['time'].max():.1f} seconds")
```

## analyze telemetry and build car model and then determine ideal racing line

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter
from scipy.stats import gaussian_kde
from sklearn.cluster import DBSCAN
import seaborn as sns

def analyze_car_performance(lap_data_file):
    """
    Analyze multiple laps of car telemetry data to determine performance characteristics.
    
    Parameters:
    -----------
    lap_data_file: CSV file containing lap telemetry data
    
    Returns:
    --------
    dict: Performance metrics and parameters
    """
    # Load the telemetry data
    print(f"Loading telemetry data from {lap_data_file}...")
    df = pd.read_csv(lap_data_file)
    
    print(f"Loaded {len(df)} data points across {df['lap'].nunique()} laps")
    
    # Use the exact column names from your dataset
    speed_column = 'speed'  # Raw speed (needs conversion to mph)
    accel_long_col = 'accel_longitudinal'
    accel_lat_col = 'accel_lateral'
    turn_rate_col = 'turn_rate'
    
    print(f"Using columns: speed='{speed_column}', longitudinal acceleration='{accel_long_col}', "
          f"lateral acceleration='{accel_lat_col}', turn rate='{turn_rate_col}'")
    
    # Basic cleaning and preprocessing
    # Convert to numeric if not already
    numeric_cols = [speed_column, accel_long_col, accel_lat_col, turn_rate_col, 'g_force', 'g_force_smoothed']
    
    for col in numeric_cols:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors='coerce')
    
    # Convert speed to mph if it's not already (assuming it might be in m/s)
    # Check if speed is likely in m/s (most speeds would be under 50 m/s)
    if df[speed_column].mean() < 50:
        print("Speed appears to be in m/s, converting to mph...")
        df['speed_mph'] = df[speed_column] * 2.23694  # m/s to mph
    else:
        # Already in mph
        df['speed_mph'] = df[speed_column]
    
    print(f"Speed range: {df['speed_mph'].min():.1f} - {df['speed_mph'].max():.1f} mph")
    
    # Filter out anomalies and stationary points
    df = df[df['speed_mph'] > 5].copy()  # Filter out very slow/stopped points
    
    # 1. Extract maximum speed
    max_speed = df['speed_mph'].quantile(0.995)  # Use 99.5 percentile for robustness
    
    # 2. Extract maximum acceleration (longitudinal)
    # Positive values are acceleration, negative are braking
    max_accel = df[accel_long_col].quantile(0.995)
    max_decel = abs(df[accel_long_col].quantile(0.005))  # Use abs to get positive value
    
    # 3. Extract maximum lateral acceleration (cornering capability)
    # Take absolute value since lateral accel can be positive or negative
    df['abs_lateral_accel'] = df[accel_lat_col].abs()
    max_lateral_accel = df['abs_lateral_accel'].quantile(0.995)
    
    # 4. Analyze speed vs. curvature relationship (cornering performance)
    # Calculate curvature from turn rate and speed
    # Avoid division by zero or very small values
    df['curvature'] = df[turn_rate_col] / df['speed_mph'].clip(lower=1.0)
    
    # Filter out extreme curvature values
    low, high = df['curvature'].quantile([0.01, 0.99])
    df_curve = df[(df['curvature'] > low) & (df['curvature'] < high)].copy()
    
    # Group by curvature bins and find max speeds in each bin
    bins = np.linspace(df_curve['curvature'].min(), df_curve['curvature'].max(), 20)
    df_curve['curvature_bin'] = pd.cut(df_curve['curvature'], bins)
    
    max_speed_by_curvature = df_curve.groupby('curvature_bin')['speed_mph'].quantile(0.95)
    
    # 5. Calculate acceleration and braking performance at different speeds
    # Group by speed bins and find max/min acceleration in each bin
    speed_bins = np.linspace(0, max_speed, 20)
    df['speed_bin'] = pd.cut(df['speed_mph'], speed_bins)
    
    accel_by_speed = df.groupby('speed_bin')[accel_long_col].quantile(0.95)
    decel_by_speed = df.groupby('speed_bin')[accel_long_col].quantile(0.05)
    
    # 6. Identify the car's performance envelope
    # Create a 2D histogram of lateral vs. longitudinal acceleration
    # This creates a "traction circle" or "g-g diagram"
    g_envelope = df[[accel_long_col, accel_lat_col]].copy()
    
    # 7. Analyze best lap and sector times
    lap_times = df.groupby('lap')['lap_seconds_elapsed'].max().sort_values()
    best_lap_time = lap_times.iloc[0]
    best_lap = lap_times.index[0]
    
    # Sector times if available
    if 'sector' in df.columns and 'sector_time' in df.columns:
        sector_times = df.groupby(['sector'])['sector_time'].min()
        sum_of_best_sectors = sector_times.sum()
    else:
        sector_times = pd.Series()
        sum_of_best_sectors = None
    
    # 8. Compile performance metrics
    performance = {
        'max_speed_mph': max_speed,
        'max_acceleration_mps2': max_accel,
        'max_deceleration_mps2': max_decel,
        'max_lateral_acceleration_mps2': max_lateral_accel,
        'best_lap_time': best_lap_time,
        'best_lap': best_lap,
        'sector_times': sector_times,
        'sum_of_best_sectors': sum_of_best_sectors,
        'speed_curvature_relationship': max_speed_by_curvature,
        'acceleration_by_speed': accel_by_speed,
        'deceleration_by_speed': decel_by_speed,
        'performance_envelope': g_envelope,
        'speed_column': speed_column,
        'accel_long_col': accel_long_col,
        'accel_lat_col': accel_lat_col
    }
    
    print(f"Analysis complete. Extracted performance metrics from {df['lap'].nunique()} laps.")
    
    # Print key metrics
    print(f"\nCar Performance Summary:")
    print(f"------------------------")
    print(f"Maximum Speed: {max_speed:.1f} mph")
    print(f"Maximum Acceleration: {max_accel:.2f} m/s²")
    print(f"Maximum Deceleration (braking): {max_decel:.2f} m/s²")
    print(f"Maximum Lateral Acceleration: {max_lateral_accel:.2f} m/s²")
    print(f"Best Lap Time: {best_lap_time:.2f} seconds (Lap {best_lap})")
    if sum_of_best_sectors is not None:
        print(f"Sum of Best Sectors: {sum_of_best_sectors:.2f} seconds")
        print(f"Theoretical Improvement: {best_lap_time - sum_of_best_sectors:.2f} seconds")
    
    return performance

def create_car_model(performance):
    """
    Create a simplified car model based on performance metrics.
    This can be used in racing line optimization.
    """
    car_model = {
        'max_speed_mph': performance['max_speed_mph'],
        'max_accel_mps2': performance['max_acceleration_mps2'],
        'max_decel_mps2': performance['max_deceleration_mps2'],
        'max_lateral_accel_mps2': performance['max_lateral_acceleration_mps2'],
    }
    
    # Create speed-dependent acceleration model
    accel_by_speed = performance['acceleration_by_speed']
    speed_values = [x.mid for x in accel_by_speed.index]
    accel_values = accel_by_speed.values
    
    # Create an interpolation function
    from scipy.interpolate import interp1d
    if len(speed_values) > 1:
        car_model['accel_function'] = interp1d(
            speed_values, accel_values, bounds_error=False, 
            fill_value=(accel_values[0], accel_values[-1])
        )
    
    # Create speed-dependent deceleration model
    decel_by_speed = performance['deceleration_by_speed']
    decel_values = decel_by_speed.values
    
    if len(speed_values) > 1:
        car_model['decel_function'] = interp1d(
            speed_values, decel_values, bounds_error=False,
            fill_value=(decel_values[0], decel_values[-1])
        )
    
    return car_model

def apply_car_model_to_racing_line(df_racing_line, car_model):
    """
    Apply the car model to a racing line to get realistic speeds.
    """
    # Get the distances between points
    lats = df_racing_line['latitude'].values
    lons = df_racing_line['longitude'].values
    
    distances = []
    for i in range(1, len(lats)):
        d = haversine_distance(lats[i-1], lons[i-1], lats[i], lons[i])
        distances.append(d)
    distances.insert(0, distances[0] if distances else 0)
    
    # Convert mph to m/s for physics calculations
    max_speed = car_model['max_speed_mph'] * 0.44704
    
    # Get the curvature values
    curvature = df_racing_line['curvature'].values
    
    # Calculate maximum cornering speed using v² = a*r
    # where a is lateral acceleration and r is radius (1/curvature)
    max_lateral_accel = car_model['max_lateral_accel_mps2']
    cornering_speeds_ms = np.sqrt(max_lateral_accel / np.maximum(curvature, 1e-10))
    
    # Cap at max speed
    cornering_speeds_ms = np.minimum(cornering_speeds_ms, max_speed)
    
    # Now implement forward and backward passes with our acceleration model
    forward_speeds = np.zeros_like(cornering_speeds_ms)
    backward_speeds = np.zeros_like(cornering_speeds_ms)
    
    # Pass 1: Forward - start with min speed and accelerate as allowed
    forward_speeds[0] = 5 * 0.44704  # Starting at 5 mph
    
    for i in range(1, len(forward_speeds)):
        # Get appropriate acceleration for current speed
        current_speed_mph = forward_speeds[i-1] / 0.44704
        
        if 'accel_function' in car_model:
            accel = car_model['accel_function'](current_speed_mph)
        else:
            accel = car_model['max_accel_mps2']
        
        # v² = u² + 2as
        distance = distances[i]
        prev_speed = forward_speeds[i-1]
        max_speed_achievable = np.sqrt(prev_speed**2 + 2 * accel * distance)
        
        # Limit to cornering speed
        forward_speeds[i] = min(max_speed_achievable, cornering_speeds_ms[i])
    
    # Pass 2: Backward - ensure we can brake in time
    backward_speeds[-1] = forward_speeds[-1]  # End speed
    
    for i in range(len(backward_speeds)-2, -1, -1):
        # Get appropriate deceleration for current speed
        current_speed_mph = backward_speeds[i+1] / 0.44704
        
        if 'decel_function' in car_model:
            decel = abs(car_model['decel_function'](current_speed_mph))  # Make positive
        else:
            decel = car_model['max_decel_mps2']
        
        # v² = u² - 2as (rearranged for u)
        distance = distances[i+1]
        next_speed = backward_speeds[i+1]
        
        # Avoid negative values under the sqrt
        discriminant = next_speed**2 + 2 * decel * distance
        if discriminant > 0:
            max_starting_speed = np.sqrt(discriminant)
        else:
            max_starting_speed = next_speed  # Fallback if calculation fails
            
        # Take minimum to respect acceleration limits
        backward_speeds[i] = min(max_starting_speed, forward_speeds[i])
    
    # The final speeds respect both cornering limits and acceleration/deceleration
    final_speeds_ms = backward_speeds
    
    # Convert back to mph for output
    final_speeds_mph = final_speeds_ms / 0.44704
    
    # Add to the DataFrame
    df_racing_line['speed_mph'] = final_speeds_mph
    
    # Calculate lap time based on speeds and distances
    # Time = distance / average_speed for each segment
    segment_times = []
    for i in range(1, len(final_speeds_ms)):
        # Average speed over the segment
        avg_speed = (final_speeds_ms[i-1] + final_speeds_ms[i]) / 2
        if avg_speed > 0:
            time = distances[i] / avg_speed
        else:
            time = 0  # Avoid division by zero
        segment_times.append(time)
    
    # Calculate cumulative time
    cumulative_time = np.cumsum([0] + segment_times)
    df_racing_line['time'] = cumulative_time
    
    # Calculate acceleration between points
    accelerations = np.zeros_like(final_speeds_ms)
    for i in range(1, len(final_speeds_ms)):
        # v - u / t
        speed_diff = final_speeds_ms[i] - final_speeds_ms[i-1]
        time_diff = segment_times[i-1]
        if time_diff > 0:
            accelerations[i] = speed_diff / time_diff
        else:
            accelerations[i] = 0
    
    df_racing_line['acceleration_mps2'] = accelerations
    
    return df_racing_line

def analyze_lap_consistency(df):
    """
    Analyze the consistency across multiple laps to identify areas of improvement.
    """
    # Group data by lap
    laps = df['lap'].unique()
    
    if len(laps) < 2:
        print("Need at least 2 laps for consistency analysis")
        return None
    
    # Calculate lap times
    lap_times = df.groupby('lap')['lap_seconds_elapsed'].max().sort_values()
    
    # Identify the best lap
    best_lap = lap_times.idxmin()
    
    print(f"Best lap: {best_lap} with time {lap_times.min():.2f} seconds")
    print(f"Average lap time: {lap_times.mean():.2f} seconds")
    print(f"Lap time standard deviation: {lap_times.std():.2f} seconds")
    
    # Extract the best lap data
    best_lap_data = df[df['lap'] == best_lap].copy()
    
    # Calculate distance for each lap if not already present
    if 'distance' not in df.columns:
        print("Calculating distance column for each lap...")
        
        # For each lap, calculate cumulative distance
        all_laps_with_distance = []
        
        for lap_num in laps:
            lap_data = df[df['lap'] == lap_num].copy().sort_values('lap_seconds_elapsed')
            
            # Calculate distance using latitude/longitude
            lats = lap_data['latitude'].values
            lons = lap_data['longitude'].values
            
            # Calculate distances between consecutive points
            distances = []
            for i in range(1, len(lats)):
                d = haversine_distance(lats[i-1], lons[i-1], lats[i], lons[i])
                distances.append(d)
            
            # Add cumulative distance column
            lap_data['distance'] = np.insert(np.cumsum(distances), 0, 0)
            all_laps_with_distance.append(lap_data)
        
        # Combine all laps back together
        df = pd.concat(all_laps_with_distance)
    
    # First, normalize distance to 0-100% of lap
    # Group by lap and calculate normalized distance for each lap separately
    df_with_norm = []
    
    for lap_num in laps:
        lap_data = df[df['lap'] == lap_num].copy()
        lap_dist_max = lap_data['distance'].max()
        lap_dist_min = lap_data['distance'].min()
        
        # Avoid division by zero
        if lap_dist_max > lap_dist_min:
            lap_data['normalized_distance'] = (lap_data['distance'] - lap_dist_min) / (lap_dist_max - lap_dist_min) * 100
        else:
            lap_data['normalized_distance'] = 0
        
        df_with_norm.append(lap_data)
    
    # Combine all laps back together
    df = pd.concat(df_with_norm)
    
    # Create distance bins (every 5% of lap)
    bins = np.linspace(0, 100, 21)  # 0%, 5%, 10%, ... 100%
    df['distance_bin'] = pd.cut(df['normalized_distance'], bins)
    
    # Calculate speed statistics in each bin across all laps
    speed_stats = df.groupby('distance_bin')['speed_mph' if 'speed_mph' in df.columns else 'speed'].agg(['mean', 'std', 'min', 'max'])
    
    # Identify high-variance sections
    high_variance_sections = speed_stats[speed_stats['std'] > speed_stats['std'].mean()].sort_values('std', ascending=False)
    
    # Analyze braking consistency
    # Assuming negative acceleration values represent braking
    braking_data = df[df['accel_longitudinal'] < -0.5].copy()
    braking_data['braking_intensity'] = -braking_data['accel_longitudinal']
    
    if len(braking_data) > 0:
        braking_stats = braking_data.groupby(['lap', 'distance_bin'])['braking_intensity'].max().reset_index()
        braking_variance = braking_stats.groupby('distance_bin')['braking_intensity'].agg(['mean', 'std'])
        inconsistent_braking = braking_variance[braking_variance['std'] > braking_variance['std'].mean()]
    else:
        inconsistent_braking = pd.DataFrame()
    
    # Return analysis results
    return {
        'lap_times': lap_times,
        'best_lap': best_lap,
        'speed_consistency': speed_stats,
        'high_variance_sections': high_variance_sections,
        'braking_consistency': inconsistent_braking if len(inconsistent_braking) > 0 else None
    }

def plot_consistency_analysis(consistency_results, df):
    """
    Visualize the consistency analysis results.
    """
    # Check if normalized_distance is in df, if not, recalculate it
    if 'normalized_distance' not in df.columns:
        print("Adding normalized_distance to dataframe for plotting...")
        # For each lap, calculate normalized distance
        df_with_norm = []
        
        for lap in df['lap'].unique():
            lap_data = df[df['lap'] == lap].copy()
            
            if 'distance' not in lap_data.columns:
                # Calculate distance for this lap
                lats = lap_data['latitude'].values
                lons = lap_data['longitude'].values
                
                distances = []
                for i in range(1, len(lats)):
                    d = haversine_distance(lats[i-1], lons[i-1], lats[i], lons[i])
                    distances.append(d)
                
                lap_data['distance'] = np.insert(np.cumsum(distances), 0, 0)
            
            # Normalize distance for this lap
            lap_dist_max = lap_data['distance'].max()
            lap_dist_min = lap_data['distance'].min()
            
            if lap_dist_max > lap_dist_min:
                lap_data['normalized_distance'] = (lap_data['distance'] - lap_dist_min) / (lap_dist_max - lap_dist_min) * 100
            else:
                lap_data['normalized_distance'] = 0
            
            df_with_norm.append(lap_data)
        
        # Reassign the dataframe with the new normalized distances
        df = pd.concat(df_with_norm)
    
    # Check if speed_mph exists, if not use speed
    speed_col = 'speed_mph' if 'speed_mph' in df.columns else 'speed'
    
    plt.figure(figsize=(18, 10))
    
    # 1. Lap time distribution
    plt.subplot(2, 2, 1)
    lap_times = consistency_results['lap_times']
    sns.histplot(lap_times.values, kde=True)
    plt.axvline(x=lap_times.min(), color='r', linestyle='--', label=f'Best: {lap_times.min():.2f}s')
    plt.axvline(x=lap_times.mean(), color='g', linestyle='--', label=f'Mean: {lap_times.mean():.2f}s')
    plt.title('Lap Time Distribution')
    plt.xlabel('Lap Time (seconds)')
    plt.ylabel('Count')
    plt.legend()
    
    # 2. Speed consistency around the lap
    plt.subplot(2, 2, 2)
    speed_stats = consistency_results['speed_consistency']
    
    # Get the bin midpoints for x-axis
    x = [(interval.left + interval.right)/2 for interval in speed_stats.index]
    
    plt.plot(x, speed_stats['mean'], 'b-', label='Mean Speed')
    plt.fill_between(x, 
                     speed_stats['mean'] - speed_stats['std'], 
                     speed_stats['mean'] + speed_stats['std'], 
                     alpha=0.3, color='b',
                     label='±1 Std Dev')
    
    # Highlight high variance sections
    high_var = consistency_results['high_variance_sections']
    if len(high_var) > 0:
        high_var_x = [(interval.left + interval.right)/2 for interval in high_var.index]
        plt.scatter(high_var_x, high_var['mean'], color='r', s=50, zorder=5,
                   label='High Variance Sections')
    
    plt.title('Speed Consistency Around Lap')
    plt.xlabel('Lap Distance (%)')
    plt.ylabel('Speed (mph)')
    plt.grid(True, alpha=0.3)
    plt.legend()
    
    # 3. Lap overlay (speed vs. distance)
    plt.subplot(2, 2, 3)
    
    # Select a subset of laps to avoid cluttering
    subset_laps = sorted(df['lap'].unique())[:10]  # First 10 laps
    
    for lap in subset_laps:
        lap_data = df[df['lap'] == lap].copy()
        lap_data = lap_data.sort_values('normalized_distance')
        plt.plot(lap_data['normalized_distance'], lap_data[speed_col], 
                alpha=0.7, 
                linewidth=1 if lap != consistency_results['best_lap'] else 2,
                label=f"Lap {lap}" if lap == consistency_results['best_lap'] else None)
    
    # Make the best lap stand out
    best_lap_data = df[df['lap'] == consistency_results['best_lap']].copy()
    best_lap_data = best_lap_data.sort_values('normalized_distance')
    plt.plot(best_lap_data['normalized_distance'], best_lap_data[speed_col], 
            'r-', linewidth=2, label=f"Best Lap ({consistency_results['best_lap']})")
    
    plt.title('Speed Profiles Overlay')
    plt.xlabel('Lap Distance (%)')
    plt.ylabel('Speed (mph)')
    plt.grid(True, alpha=0.3)
    plt.legend()
    
    # 4. Braking consistency
    plt.subplot(2, 2, 4)
    
    if consistency_results['braking_consistency'] is not None:
        braking = consistency_results['braking_consistency']
        
        # Get the bin midpoints for x-axis
        x = [(interval.left + interval.right)/2 for interval in braking.index]
        
        plt.bar(x, braking['mean'], yerr=braking['std'], 
               alpha=0.7, capsize=5)
        
        plt.title('Braking Consistency')
        plt.xlabel('Lap Distance (%)')
        plt.ylabel('Braking Intensity (m/s²)')
        plt.grid(True, axis='y', alpha=0.3)
    else:
        plt.text(0.5, 0.5, 'Insufficient braking data',
                ha='center', va='center', fontsize=14)
        plt.axis('off')
    
    plt.tight_layout()
    plt.savefig('lap_consistency_analysis.png', dpi=300, bbox_inches='tight')
    print("Consistency analysis visualizations saved to 'lap_consistency_analysis.png'")

# Define haversine_distance function if not already defined
def haversine_distance(lat1, lon1, lat2, lon2):
    """Calculate the great circle distance between two points in meters"""
    # Convert decimal degrees to radians
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    
    # Haversine formula
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    r = 6371000  # Radius of earth in meters
    return c * r

# Function to apply the derived car model to optimize a racing line
def optimize_racing_line_with_car_model(centerline_file, track_file, car_model, frequency_hz=100):
    """
    Uses the car performance model to optimize a racing line with realistic speeds
    
    Parameters:
    -----------
    centerline_file: CSV file with track centerline
    track_file: CSV file with track boundaries
    car_model: Dict containing car performance parameters
    frequency_hz: Resampling frequency for the final racing line
    
    Returns:
    --------
    DataFrame containing optimized racing line with realistic speeds
    """
    # Load track data
    df_center = pd.read_csv(centerline_file)
    df_track = pd.read_csv(track_file)
    
    # Separate inner and outer boundaries
    df_inner = df_track[df_track['trace_type'] == 'inner'].copy()
    df_outer = df_track[df_track['trace_type'] == 'outer'].copy()
    
    print(f"Loaded centerline with {len(df_center)} points")
    print(f"Loaded track with {len(df_inner)} inner points and {len(df_outer)} outer points")
    
    # Optimize the racing line based on the centerline and track boundaries
    df_racing_line = optimize_racing_line(df_center, df_inner, df_outer, smoothing=True)
    print(f"Optimized racing line with {len(df_racing_line)} points")
    
    # Apply the car model to get realistic speeds
    df_racing_line = apply_car_model_to_racing_line(df_racing_line, car_model)
    
    # Resample the racing line at the desired frequency
    df_resampled = finalize_racing_line_resampling(df_racing_line, frequency_hz)
    print(f"Resampled racing line to {len(df_resampled)} points at {frequency_hz}Hz")
    
    # Print summary of the optimized racing line
    print("\nRacing Line Summary:")
    print(f"Total distance: {df_resampled['distance'].max():.1f} meters")
    print(f"Estimated lap time: {df_resampled['time'].max():.2f} seconds")
    print(f"Speed range: {df_resampled['speed_mph'].min():.1f} - {df_resampled['speed_mph'].max():.1f} mph")
    print(f"Average speed: {df_resampled['speed_mph'].mean():.1f} mph")
    
    return df_resampled

def finalize_racing_line_resampling(df_racing_line, frequency_hz=100):
    """
    Resamples the racing line at the desired frequency.
    """
    if frequency_hz <= 0:
        return df_racing_line
    
    # Ensure there's a distance column
    if 'distance' not in df_racing_line.columns:
        print("Calculating distance for racing line...")
        # Calculate cumulative distance
        lats = df_racing_line['latitude'].values
        lons = df_racing_line['longitude'].values
        
        distances = []
        for i in range(1, len(lats)):
            d = haversine_distance(lats[i-1], lons[i-1], lats[i], lons[i])
            distances.append(d)
        
        df_racing_line['distance'] = np.insert(np.cumsum(distances), 0, 0)
    
    # Get total time
    total_time = df_racing_line['time'].max()
    
    # Create equally spaced time points
    num_points = int(total_time * frequency_hz) + 1
    new_times = np.linspace(0, total_time, num_points)
    
    # Create interpolation functions for all columns
    interp_cols = ['latitude', 'longitude', 'speed_mph', 'distance', 'bearing', 
                   'curvature', 'acceleration_mps2']
    interp_funcs = {}
    
    for col in interp_cols:
        if col in df_racing_line.columns:
            interp_funcs[col] = interp1d(
                df_racing_line['time'], 
                df_racing_line[col],
                bounds_error=False, 
                fill_value='extrapolate'
            )
    
    # Create the resampled DataFrame
    resampled_data = {'time': new_times}
    
    for col, func in interp_funcs.items():
        resampled_data[col] = func(new_times)
    
    df_resampled = pd.DataFrame(resampled_data)
    
    # Add ID column for visualization
    df_resampled['id'] = np.arange(len(df_resampled))
    
    return df_resampled

def optimize_racing_line_with_car_model(centerline_file, track_file, car_model, frequency_hz=100):
    """
    Uses the car performance model to optimize a racing line with realistic speeds
    
    Parameters:
    -----------
    centerline_file: CSV file with track centerline
    track_file: CSV file with track boundaries
    car_model: Dict containing car performance parameters
    frequency_hz: Resampling frequency for the final racing line
    
    Returns:
    --------
    DataFrame containing optimized racing line with realistic speeds
    """
    # Load track data
    df_center = pd.read_csv(centerline_file)
    df_track = pd.read_csv(track_file)
    
    # Separate inner and outer boundaries
    df_inner = df_track[df_track['trace_type'] == 'inner'].copy()
    df_outer = df_track[df_track['trace_type'] == 'outer'].copy()
    
    print(f"Loaded centerline with {len(df_center)} points")
    print(f"Loaded track with {len(df_inner)} inner points and {len(df_outer)} outer points")
    
    # Optimize the racing line based on the centerline and track boundaries
    df_racing_line = optimize_racing_line(df_center, df_inner, df_outer, smoothing=True)
    print(f"Optimized racing line with {len(df_racing_line)} points")
    
    # Apply the car model to get realistic speeds
    df_racing_line = apply_car_model_to_racing_line(df_racing_line, car_model)
    
    # Resample the racing line at the desired frequency
    df_resampled = finalize_racing_line_resampling(df_racing_line, frequency_hz)
    print(f"Resampled racing line to {len(df_resampled)} points at {frequency_hz}Hz")
    
    # Ensure all required columns exist before printing summary
    required_cols = ['distance', 'time', 'speed_mph']
    missing_cols = [col for col in required_cols if col not in df_resampled.columns]
    
    if missing_cols:
        print(f"Warning: Missing columns in racing line: {missing_cols}")
        
        # Fill in missing columns with placeholders
        for col in missing_cols:
            if col == 'distance' and 'latitude' in df_resampled.columns and 'longitude' in df_resampled.columns:
                # Calculate distance if missing
                lats = df_resampled['latitude'].values
                lons = df_resampled['longitude'].values
                
                distances = []
                for i in range(1, len(lats)):
                    d = haversine_distance(lats[i-1], lons[i-1], lats[i], lons[i])
                    distances.append(d)
                
                df_resampled['distance'] = np.insert(np.cumsum(distances), 0, 0)
            elif col == 'time':
                # Add placeholder time if missing
                df_resampled['time'] = np.linspace(0, 60, len(df_resampled))  # Assume 60 seconds lap
            elif col == 'speed_mph':
                # Add placeholder speed if missing
                df_resampled['speed_mph'] = 30  # Placeholder value
    
    # Print summary of the optimized racing line
    print("\nRacing Line Summary:")
    try:
        if 'distance' in df_resampled:
            print(f"Total distance: {df_resampled['distance'].max():.1f} meters")
        if 'time' in df_resampled:
            print(f"Estimated lap time: {df_resampled['time'].max():.2f} seconds")
        if 'speed_mph' in df_resampled:
            print(f"Speed range: {df_resampled['speed_mph'].min():.1f} - {df_resampled['speed_mph'].max():.1f} mph")
            print(f"Average speed: {df_resampled['speed_mph'].mean():.1f} mph")
    except Exception as e:
        print(f"Error generating summary: {e}")
    
    return df_resampled

def compare_telemetry_to_racing_line(telemetry_file, racing_line_df):
    """
    Compare actual telemetry with the optimized racing line to identify
    areas for improvement.
    """
    # Load telemetry data
    df_telemetry = pd.read_csv(telemetry_file)
    
    # Determine which columns to use for speed
    if 'speed_mph' in df_telemetry.columns:
        speed_col_telemetry = 'speed_mph'
    elif 'speed' in df_telemetry.columns:
        # Check if speed is likely in m/s (most speeds would be under 50 m/s)
        if df_telemetry['speed'].mean() < 50:
            print("Speed in telemetry appears to be in m/s, converting to mph...")
            df_telemetry['speed_mph'] = df_telemetry['speed'] * 2.23694  # m/s to mph
            speed_col_telemetry = 'speed_mph'
        else:
            # Assume it's already in mph
            df_telemetry['speed_mph'] = df_telemetry['speed']
            speed_col_telemetry = 'speed_mph'
    else:
        raise ValueError("No speed column found in telemetry data")
    
    # Determine which column to use for racing line speed
    if 'speed_mph' in racing_line_df.columns:
        speed_col_racing = 'speed_mph'
    elif 'speed' in racing_line_df.columns:
        # Check if speed is likely in m/s
        if racing_line_df['speed'].mean() < 50:
            print("Speed in racing line appears to be in m/s, converting to mph...")
            racing_line_df['speed_mph'] = racing_line_df['speed'] * 2.23694  # m/s to mph
            speed_col_racing = 'speed_mph'
        else:
            # Assume it's already in mph
            racing_line_df['speed_mph'] = racing_line_df['speed']
            speed_col_racing = 'speed_mph'
    else:
        print("No speed column in racing line data, using default value")
        racing_line_df['speed_mph'] = 30  # Default value
        speed_col_racing = 'speed_mph'
    
    # Find the best lap in telemetry
    lap_times = df_telemetry.groupby('lap')['lap_seconds_elapsed'].max()
    best_lap = lap_times.idxmin()
    
    # Extract best lap data
    best_lap_data = df_telemetry[df_telemetry['lap'] == best_lap].copy()
    
    # Calculate lap fractions for comparison (0-100%)
    if 'distance' not in best_lap_data.columns:
        # Calculate cumulative distance
        lats = best_lap_data['latitude'].values
        lons = best_lap_data['longitude'].values
        
        distances = []
        for i in range(1, len(lats)):
            d = haversine_distance(lats[i-1], lons[i-1], lats[i], lons[i])
            distances.append(d)
            
        best_lap_data['distance'] = np.insert(np.cumsum(distances), 0, 0)
    
    # Normalize distance to 0-100%
    best_lap_data['lap_fraction'] = (best_lap_data['distance'] / best_lap_data['distance'].max()) * 100
    
    # Also normalize racing line distance if it exists
    if 'distance' not in racing_line_df.columns:
        # Calculate distance for racing line
        lats = racing_line_df['latitude'].values
        lons = racing_line_df['longitude'].values
        
        distances = []
        for i in range(1, len(lats)):
            d = haversine_distance(lats[i-1], lons[i-1], lats[i], lons[i])
            distances.append(d)
        
        racing_line_df['distance'] = np.insert(np.cumsum(distances), 0, 0)
    
    racing_line_df['lap_fraction'] = (racing_line_df['distance'] / racing_line_df['distance'].max()) * 100
    
    # Create interpolation functions for the racing line
    racing_speed_func = interp1d(
        racing_line_df['lap_fraction'],
        racing_line_df[speed_col_racing],
        bounds_error=False,
        fill_value='extrapolate'
    )
    
    # Calculate the optimal speed at each point in the actual lap
    best_lap_data['optimal_speed'] = racing_speed_func(best_lap_data['lap_fraction'])
    
    # Calculate speed difference
    best_lap_data['speed_diff'] = best_lap_data[speed_col_telemetry] - best_lap_data['optimal_speed']
    
    # Visualize the comparison
    plt.figure(figsize=(15, 10))
    
    # 1. Plot speed profiles
    plt.subplot(2, 1, 1)
    plt.plot(best_lap_data['lap_fraction'], best_lap_data[speed_col_telemetry], 'b-', label='Actual Speed')
    plt.plot(racing_line_df['lap_fraction'], racing_line_df[speed_col_racing], 'r-', label='Optimal Speed')
    
    plt.title(f'Speed Comparison - Actual vs. Optimal Racing Line (Lap {best_lap})')
    plt.xlabel('Lap Distance (%)')
    plt.ylabel('Speed (mph)')
    plt.grid(True, alpha=0.3)
    plt.legend()
    
    # 2. Plot speed difference
    plt.subplot(2, 1, 2)
    plt.bar(best_lap_data['lap_fraction'], best_lap_data['speed_diff'], width=0.5, alpha=0.7)
    plt.axhline(y=0, color='k', linestyle='-', linewidth=1)
    
    # Highlight areas with significant differences
    threshold = 5  # mph
    too_slow = best_lap_data[best_lap_data['speed_diff'] < -threshold]
    too_fast = best_lap_data[best_lap_data['speed_diff'] > threshold]
    
    plt.scatter(too_slow['lap_fraction'], too_slow['speed_diff'], color='r', s=30, label='Too Slow')
    plt.scatter(too_fast['lap_fraction'], too_fast['speed_diff'], color='g', s=30, label='Too Fast')
    
    plt.title('Speed Difference (Actual - Optimal)')
    plt.xlabel('Lap Distance (%)')
    plt.ylabel('Speed Difference (mph)')
    plt.grid(True, alpha=0.3)
    plt.legend()
    
    plt.tight_layout()
    plt.savefig('racing_line_comparison.png', dpi=300, bbox_inches='tight')
    print("Racing line comparison saved to 'racing_line_comparison.png'")
    
    # Identify major improvement areas
    improvement_areas = pd.DataFrame()
    
    # Group by 5% sections of the lap
    best_lap_data['section'] = pd.cut(best_lap_data['lap_fraction'], bins=np.linspace(0, 100, 21))
    section_stats = best_lap_data.groupby('section')['speed_diff'].mean().reset_index()
    
    # Find the sections with largest average speed deficit
    slow_sections = section_stats[section_stats['speed_diff'] < -threshold].sort_values('speed_diff')
    
    if len(slow_sections) > 0:
        print("\nAreas for Improvement (Driver too slow):")
        for i, (_, row) in enumerate(slow_sections.iterrows()):
            section = row['section']
            diff = row['speed_diff']
            print(f"{i+1}. Track section {section.left:.0f}%-{section.right:.0f}%: " 
                  f"Average {-diff:.1f} mph too slow")
    
    return best_lap_data

# Execute if run directly
if __name__ == "__main__":
    telemetry_file = "telemetry.csv"  # Update with your file path
    centerline_file = "centerline.csv"
    track_file = "track.csv"
    
    # Analyze telemetry and get car model
    car_model = analyze_telemetry(telemetry_file)
    
    # Save car model parameters for use in racing line optimization
    import json
    with open('car_model.json', 'w') as f:
        # Convert any numpy floats to Python floats for JSON serialization
        serializable_model = {k: float(v) if isinstance(v, np.float64) else v 
                             for k, v in car_model.items() 
                             if k not in ['accel_function', 'decel_function']}
        json.dump(serializable_model, f, indent=2)
    
    print("Car model saved to 'car_model.json'")
    
    # Optimize racing line using the car model
    df_racing_line = optimize_racing_line_with_car_model(
        centerline_file, track_file, car_model, frequency_hz=100
    )
    
    # Save the optimized racing line
    output_file = 'gateway_racing_line_100hz.csv'
    df_racing_line.to_csv(output_file, index=False)
    print(f"Optimized racing line saved to '{output_file}'")
    
    # Compare the optimized line with actual telemetry
    comparison = compare_telemetry_to_racing_line(telemetry_file, df_racing_line)
    
    print("\nAnalysis complete!")
```
