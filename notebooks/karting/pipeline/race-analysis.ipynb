```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
# Replace 'database.db' with your actual database file path
db_path = raw_gateway[0]['filename']
conn = sqlite3.connect(db_path)

# Query to get the first row of the Location table
query = "SELECT * FROM Location LIMIT 1;"

try:
    # Execute query and fetch the first row
    first_row = pd.read_sql_query(query, conn)
    
    # Display the first row
    if not first_row.empty:
        print("First row of the Location table:")
        display(first_row)  # This is a Jupyter notebook function to display dataframes nicely
        
        # Alternatively, you can print column by column:
        print("\nColumn by column:")
        for column in first_row.columns:
            print(f"{column}: {first_row[column].iloc[0]}")
    else:
        print("The Location table is empty.")
        
except sqlite3.OperationalError as e:
    if "no such table" in str(e).lower():
        print("Error: The Location table does not exist in this database.")
    else:
        print(f"SQLite error: {e}")

# Close the connection when done
conn.close()
```

```python
import sqlite3
import pandas as pd
from datetime import datetime

for gateway in raw_gateway:
    db_path = gateway['filename']
    conn = sqlite3.connect(db_path)
    
    query = "SELECT * FROM Location LIMIT 1;"
    first_row = pd.read_sql_query(query, conn)
    
    if not first_row.empty:
        utc_timestamp = first_row['utc_time'].iloc[0]
        readable_time = datetime.fromisoformat(utc_timestamp.replace('Z', '+00:00')).strftime('%Y-%m-%d %H:%M')
        print(f"{gateway['filename']}: {readable_time} Visit {gateway['visit']} Race {gateway['race']}")
    
    conn.close()
```
